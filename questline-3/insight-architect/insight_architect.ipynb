# Insight Architect

This notebook performs exploratory data analysis on the cleaned student performance dataset using Python visualizations to uncover patterns and relationships.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../../questline-2/from-chaos-to-clarity/data-refinery/cleaned_students_performance.csv')

print('Dataset shape:', df.shape)
print(df.head())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Attendance'], bins=8, kde=True, color='steelblue')
plt.title('Distribution of Attendance')
plt.xlabel('Attendance (%)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
pass_counts = df['Pass_Status'].value_counts()
plt.figure(figsize=(6, 5))
sns.barplot(x=pass_counts.index, y=pass_counts.values, palette='Set2')
plt.title('Pass/Fail Distribution')
plt.xlabel('Outcome')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df[['Math_Score', 'Science_Score', 'English_Score']], orient='h', palette='pastel')
plt.title('Score Distribution Across Subjects')
plt.xlabel('Score')
plt.ylabel('Subject')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Attendance', y='Math_Score', hue='Pass_Status', s=100, palette='deep')
plt.title('Attendance vs Math Score')
plt.xlabel('Attendance (%)')
plt.ylabel('Math Score')
plt.tight_layout()
plt.show()
```

```python
corr = df[['Age', 'Math_Score', 'Science_Score', 'English_Score', 'Attendance']].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numeric Features')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
gender_avg = df.groupby('Gender')[['Math_Score', 'Science_Score', 'English_Score']].mean()
gender_avg.plot(kind='bar', figsize=(8, 5), alpha=0.85)
plt.title('Average Subject Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.legend(title='Subject')
plt.tight_layout()
plt.show()
```

## Insights

1. Attendance is concentrated in the high range, suggesting most students are consistently present and this may support stronger academic performance.
2. The pass/fail distribution shows a majority of students passed, indicating a generally healthy academic outcome in the dataset.
3. The box plot suggests Math and Science scores are slightly more variable than English, meaning performance is less uniform across those subjects.
4. Attendance and Math score show a positive relationship, with higher attendance generally associated with higher math performance.
5. The heatmap indicates moderate positive correlations between subject scores and attendance, suggesting academic performance is linked to consistent engagement.
6. Average subject scores by gender do not show a major difference, suggesting that gender is not a strong determinant of performance in this sample.
7. The dataset is small but shows clear patterns: strong attendance and solid subject performance tend to align with pass outcomes.
