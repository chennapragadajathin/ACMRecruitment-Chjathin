# Data Refinery

This notebook cleans a student performance dataset by handling missing values, removing duplicates, and verifying the final dataset is clean.

```python
import pandas as pd

df = pd.read_csv('../../questline-2/from-chaos-to-clarity/data-refinery/students_performance.csv')

print('Original shape:', df.shape)
print('Original duplicate rows:', df.duplicated().sum())
print('Original missing values:\n', df.isnull().sum())

df = df.drop_duplicates().copy()

numeric_cols = ['Age', 'Math_Score', 'Science_Score', 'English_Score', 'Attendance']
for col in numeric_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

df['Name'] = df['Name'].fillna('Unknown')
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Pass_Status'] = df['Pass_Status'].fillna(df['Pass_Status'].mode()[0])

print('After cleaning shape:', df.shape)
print('Duplicate rows after cleaning:', df.duplicated().sum())
print('Missing values after cleaning:\n', df.isnull().sum())

assert df.duplicated().sum() == 0
assert df.isnull().sum().sum() == 0

df.to_csv('cleaned_students_performance.csv', index=False)
print('Saved cleaned dataset to cleaned_students_performance.csv')
```
